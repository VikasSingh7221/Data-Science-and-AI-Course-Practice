1. The Smart Attendance Logger
    - The Goal: Manage a classroom attendance list and ensure no duplicate entries are recorded if a student taps their ID
card twice.
- Concepts: set , function , while.
- The Problem: Create a function mark_attendance() that keeps asking for student names until the user
types "STOP". Store names in a Set to automatically prevent duplicates.
- Input: Multiple strings (names) entered one by one.
- Output: The final count of unique students and the set itself.

```python
total_students = set()
def mark_attendance():
    while True:
        name = input("Enter the name :")
        if name.lower().strip() == "stop":
            break
        total_students.add(name)

mark_attendance()
mark_attendance()
mark_attendance()
print(f"Total Number of students : {len(total_students)}")
for i,name in enumerate(total_students, start = 1):
    print(f"{i}. {name.capitalize()}")        
```

2. The Restaurant Bill Splitter

    - The Goal: Calculate how much each person owes, including a tax and a custom tip percentage.

    - Concepts: Functions, default values, lists.

    - The Problem: Write a function split_bill(items_prices, people, tax=0.05, tip=0.10).

    - Input: A list like [25.0, 15.0, 10.0] and an integer for people.
    - Output: A single float (the amount per person).

```python
def split_bill(items_price, people, tax = 0.05, tip = 0.10):
    total = sum(items_price)

    total_tax = total * tax
    total_tip = total * tip 
    
    total_bill = total + total_tax + total_tip 
    return total_bill/ people 


people = int(input("Enter total number of people : "))
items_price = []
while True:
    item = input("Enter items")
    if item.lower().strip() == "over":
        break
    items_price.append(int(item))

print(f"Total number of people : {people}")
print(f"Total items : {items_price}")
print(f"Each person will get : {split_bill(items_price=items_price, people= people)}")
```

### **3. The Digital Wardrobe Organizer**

**The Goal:**  
Check if a specific outfit is available or if parts of it are missing.

**Concepts:**  
Multiple `if` statements, `list`, `input()`.

**The Problem:**  
You have a list: `wardrobe = ["Blue Jeans", "White Shirt", "Black Jacket"]`.  
1. Ask the user for **3 items** they want to wear.  
2. Check each item individually against the list.  

**Input:**  
Three separate `input()` strings (e.g., "Red Tie", "White Shirt", "Socks").

**Output:**  
A "Missing" warning for every item not found in the list.

**Hint:**  
Use the `not in` operator to identify missing items.

```python
wardrobe = ["Blue Jeans", "White Shirt", "Black Jacket"]

while True:
    outfit = input("Enter item (or 'over' to stop): ").strip()
    
    if outfit.lower() == "over":
        break 
    
    # Create a lowercase version of the list for comparison
    if outfit.lower() not in [item.lower() for item in wardrobe]:
        print(f"Warning !! Missing : {outfit}")
```

4. The E-Commerce Discount Engine
    - The Goal: Apply different discount rates based on the total purchase amount.
    - Concepts: if-elif-else , functions.
    - The Problem: Create a function apply_discount(total).
        - If total > $500: 20% off.,,,
        - If total > $200: 10% off...
        - Else: No discount.
    - Input: A single number (the total bill).
    - Output: The final price after the discount is subtracted.

```python
def apply_discount(total):
    if total >= 500:
        total -= total * 0.20
    elif total < 500 and total >= 200:
        total -= total * 0.10
    else:
        return "No discount."
    
    return total

while True:
    total = int(input("Enter total amount: "))
    if total == 0:
        break 
    print(f"Total : {total}, After Discount : {apply_discount(total)}")

```

7. The ATM Pin Validator
- The Goal: Lock the user out after 3 incorrect attempts.
- Concepts: while , if-else , break .
- The Problem: Set correct_pin ="1234" . Allow the user 3 tries to enter the right PIN.
- Input: Up to 3 string inputs.
- Output: "Welcome" on success, or "Card Blocked" after the 3rd fail.

```python
correct_pin = "15150109"
attempts = 0
while True:
    if attempts == 3:
        print("Card Blocked !!")
        break

    pin = input("Enter pin : ").strip()
    if pin == correct_pin:
        print("Welcome, Your pin is correct !!")
    else:
        print(f"{pin} is Incorrect pin !!!")
    attempts += 1
```
