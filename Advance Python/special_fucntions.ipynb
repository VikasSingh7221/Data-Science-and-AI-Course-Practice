## filter, zip, reduce , map, list comprehension

```python
def is_even(n):
    return n%2 == 0
```

```python
data = list(range(1, 200, 3))
print(data)
```

```python
print(list(filter(lambda n : n % 2 == 0, data)))
```

```python
print(list(map(lambda n : n % 2, data)))
```
