## File Operations


```python
import os

path = "./data/file.txt"

# Check path exists
if os.path.exists(path):
    print("Path exists")
else:
    print("Path does not exist")
```

```python
import os

path = "data/test.txt"

if os.path.isfile(path):
    print("File exists")
else:
    print("File does not exist")
```

```python
import os

path = "data"

if os.path.isdir(path):
    print("Folder exists")
else:
    print("Folder does not exist")
```

```python
from pathlib import Path

p = Path("data/test.txt")

p.exists()    # file or folder
p.is_file()   # only file
p.is_dir()    # only folder
```

```python
with open('data/file.txt','w') as file:
    for i in range(1,11):
        file.write(f"19 X {i} = {19*i}\n")
```

```python
f = open('data/file.txt')
while True:
    line = f.readline()
    if not line:
        break 
    print(line)
```

```python
f = open('data/file.txt')
while True:
    line = f.readline()
    if not line:
        break 
    print(line.strip())
```

```python
f = open('data/file.txt')
while True:
    line = f.readline()
    if line:
        print(line.strip())
    else:
        break
f.close()
```

```python
# automatically handle close openration by context manager.
with open('data/file.txt') as file:
    while True:
      line = file.readline()
      if line:
        print(line.strip())
      else:
        break
```

# Advance File Operations 



## f.tell()  : give current position of file pointer.

```python
with open('data/test_1.txt') as file:
    print(file.tell())
    all_data = file.read()
    print(file.tell())
    print(all_data)
```
