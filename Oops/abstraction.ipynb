### Abstratction : Make a contract that will follow by all the class who will use this.
- a class structure is created and all the related things have to use same number of methods.

```python
from abc import  ABC, abstractmethod


class Authentication(ABC):
    @abstractmethod
    def password_check(self, password):
        pass

    @abstractmethod
    def otp_check(self, otp):
        pass
```

### Here we have to implement all the abstract methods

```python
class web(Authentication):

    def password_check(self, password):
        if password == "":
            return False
        if password == "123456":
            return True
        return False
```

```python
w1 = web()
```

### Class with both methods

```python
class web(Authentication):

    def password_check(self, password):
        if password == "":
            return False
        if password == "123456":
            return True
        return False

    def otp_check(self, otp):
        if otp == "":
            return False
        if otp == "123456":
            return True
        return False
```

```python
w1 = web()
```

```python
if w1.password_check("123456") and w1.otp_check("123456"):
    print("Successfully Authenticated")
else:
    print("Authentication Failed")
```
