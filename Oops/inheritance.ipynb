# Inheritance

### Single level Inheritance

```python
class Device:
    def switch_on(self):
        print("Switching on")

class Tv(Device):
    def play_channel(self):
        print("Playing channel")
```

#### using object of child to access function of parent

```python
t1 = Tv()

t1.switch_on()
t1.play_channel()
```

#### using object of parent to access function of child

```python
d1 = Device()

d1.switch_on()
d1.play_channel() ## error
```

### Multi level Inheritance

```python
class Device:
    def f1(self):
        print("Inside Device")

class Company(Device):
    def f2(self):
        print("Inside Company")

class Tv_1(Company):
    def f3(self):
        print("Inside Tv_1")


t1 = Tv_1()

t1.f1()
t1.f2()
t1.f3()
```

## Multiple Inheritance

```python
class A:
    def f1(self):
        print("Inside A")

class B:
    def f2(self):
        print("Inside B")

class c(A, B):
    def f3(self):
        print("Inside c")

c1 = c()
c1.f1()
c1.f2()
c1.f3()
```

```python
class A:
    def __init__(self):
        print("Inside A")

class B:
    def __init__(self):
        print("Inside B")

class c(A, B):
    def __init__(self):
        print("Inside c")

c1 = c()
```

### Order is first A then B

```python
class A:
    def __init__(self):
        print("Inside A")

class B:
    def __init__(self):
        print("Inside B")

class c(A, B):
    def __init__(self):
        super().__init__()
        print("Inside c")

c = c()
```

### Order is first B then A

```python
class A:
    def __init__(self):
        print("Inside A")

class B:
    def __init__(self):
        print("Inside B")

class c(B, A):
    def __init__(self):
        super().__init__()
        print("Inside c")


c = c()
```

### Method Overriding

- Method get override in child class

```python
class Animal:
    def speak(self):
        print("Animal Speak")


class Dog(Animal):
    def speak(self):
        print("Dog Speak")

d1 = Dog()
d1.speak()
```

```python
d1 = Animal()
d1.speak()
```

```python
class Animal:
    def speak(self):
         print("Animal Speak")


class Dog(Animal):
    def speak(self):
        super().speak()  ## Here super() refers to parent class and converted to Animal().speak()
        print("Dog Speak")
```

```python
d = Dog()
d.speak()
```
