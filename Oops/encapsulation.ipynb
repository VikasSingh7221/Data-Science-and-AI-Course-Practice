### Encapsulation

```python
class Bank:

    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance


    def get_balance(self):
        print("Your Balance is : ", self.__balance)
```

```python
b1 = Bank("Isha", 10000)
b1.get_balance() # here it uses "_Bank__balance" inside
```

```python
b1.name
```

### Here it had added a new variable , original private value is unchanged and can be referenced by "_Bank__balance"

```python
b1.__balance
```

```python
dir(b1) ## no __balance
```

```python
b1.__balance = 100
```

```python
dir(b1)  # "__balance created"
```

### In python , private is accessed using _Bank__balance so we will learn property decorator.
- getters and setters with property decorator.
