# Getters and Setters

```python
from rich import print
```

```python
class Account:
    def __init__(self):
        self.account_balance = 1000


    @property
    def balance(self):
        return self.account_balance
    
    @balance.setter
    def balance(self, amount):
        self.account_balance = amount

    
```

```python
a = Account()
```

```python
print(a.balance, end = ' ')
a.balance = 2000
print(a.balance, end = ' ')
```
