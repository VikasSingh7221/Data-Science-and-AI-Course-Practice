### Special Methods

```python
class Demo:

    def __init__(self,a ,b):
        self.a = a 
        self.b = b


    def __str__(self):
        return f"({self.a} + i{self.b})"


    def __repr__(self):
        return f"({self.a} + i{self.b})"



d = Demo(10, 20)
```

```python
print(d)
```

## Operator Overloading

```python
class Number:
    def __init__(self, a):
        self.a = a


    def __add__(self, b):
        if not isinstance(b, str):
            return self.a + b.a

        return f"{b} is an str object"


    
```

```python
n1  = Number(10)
n2  = Number(20)
```
