```python
data = "my Name is Vikas."

print(f"{data} -> upper -> {data.upper()}")
print(f"{data} -> lower -> {data.lower()}")
print(f"{data} -> title -> {data.title()}")
print(f"{data} -> capitalize -> {data.capitalize()}")
print(f"{data} -> swapcase -> {data.swapcase()}")
```

```python
# .find("char or word") : for first available, return index from left else return -1,  case sensistive
# -1 is not a index it is like if not present.

print(f"'Isha Singh'.find'I') : {'Isha Singh'.find('I')}")
print(f"'Isha Is Singh'.find'Is') : {'isha Is Singh'.find('Is')}")
print(f"'Isha Singh'.find'Is') : {'Isha Singh'.find('Is')}")
print(f"'Isha Singh'.find'is') : {'Isha Singh'.find('is')}") # should be same. 
```

```python
# .split(sep = ) returns a list.
print(f"'Isha Singh is a Beautiful girl.'.split() ==> {'Isha Singh is a Beautifle girl.'.split()}")
print(f"'Isha/Singh/is/a/Beautiful/girl.'.split() ==> {'Isha/Singh/is/a/Beautiful/girl.'.split('/')}")
print(f"'Isha/Singh is/a/Beautiful/girl.'.split() ==> {'Isha/Singh is/a/Beautiful/girl.'.split()}")
```

```python
# .join(sep = '') returns string.
print(f"'Isha Singh is a Beautiful girl.'.join() ==> {'#'.join(['Isha' ,'Singh', 'is' ,'a', 'Beautifle', 'girl'])}")
print(f"'Isha Singh is a Beautiful girl.'.join() ==> {' '.join(['Isha' ,'Singh', 'is' ,'a', 'Beautifle', 'girl'])}")
```
