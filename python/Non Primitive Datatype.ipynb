### List Methods.
- data.pop() # when we want to remove some ele via index  deault : last
- data.remove("ele") # ele is element which we want to delete
- data.index("ele")  # returns index of ele.
- data.append("ele") # insert single ele at last
- data.insert(0,"ele") # at 0 index , insert ele

# Dictionary

- Non primitive datatype where data is stored in key-value pair.
- data stored with labels known as keys instead of positions(index).
- key must be either number , strings , tuples (but must be non mutable) and always be unique.
- Value can be anything.
- data = dict() or {}

```python
data = dict([1,2,3,4])
```

```python
### Methods of dict datatype.

data = {
    "Vikas": [10,20,30,40],
    "Isha": [100,200,300,400]
}
```

```python
print(f"data.keys() => {data.keys()}")
print(f"data.values() => {data.values()}")
print(f"data.items() => {data.items()}")
```

```python
## get() function.
print(data['Isha'])
```

```python
print(data.get('Isha','Value Not present'))
```

```python
print(data.get('Ishaa','Value Not present')) #ishaa is not there hence default value.
```

```python
data_of_tuples = [('Vikas',[10,20,30]),('Isha',[40,50,60])]
print(dict(data_of_tuples))
```

## Immutable V/s Mutable Datatype.

- Immutable - something that cannot be changed (tuple).
- mutable - something that can be changed (list).

### Tuple
- list with immutable behaviour.
- no append, no pop, no reassignment of data.


```python
data = (10,20,30)
data[0]
```

```python
data[0] = 100
```

```python
data = 10
data1 = 10,
data2 = ()
data3 = (10,)
```

```python
print(f"{data} -> {type(data)}")
print(f"{data1} -> {type(data1)}")
print(f"{data2} -> {type(data2)}")
print(f"{data3} -> {type(data3)}")
```
